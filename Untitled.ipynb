```python
import pandas as pd
import numpy as np
import time

print("=====================================")
print(" BAXTER BRAIN: WEBULL ALGO ENGINE    ")
print("=====================================")

# 1. Simulate Raw Webull API Market Data Ingestion
raw_data = {
    'ticker': ['SPY', 'BTC', 'NVDA', 'TSLA'],
    'raw_price': [546.20, 64210.00, 126.40, 198.50],
    'momentum_indicator': [0.85, -0.15, 0.92, 0.45],
    'market_noise_ratio': [0.15, 0.40, 0.10, 0.25]
}
df = pd.DataFrame(raw_data)
print("[*] Raw Webull market data ingested into pandas dataframe...")
time.sleep(1)

# 2. Apply the 888 Hz Grid Harmonization
def harmonize_888(dataframe):
    print("[*] Applying 888 Hz Grid harmonization to strip market latency...")
    time.sleep(1)
    # Zero out the noise ratio and calculate a clean target price
    dataframe['market_noise_ratio'] = 0.00
    dataframe['harmonized_price'] = round(dataframe['raw_price'] * (1 + (dataframe['momentum_indicator'] * 0.005)), 2)
    return dataframe

clean_df = harmonize_888(df.copy())
print("[+] Market noise reduced to 0.00%. Data stream is pristine.\n")
time.sleep(1)

# 3. Baxter Brain Categorization Logic
def baxter_categorize(row):
    momentum = row['momentum_indicator']
    if momentum >= 0.80:
        return "ACCUMULATE (HIGH CONVICTION)"
    elif momentum >= 0.30:
        return "ACCUMULATE (MOMENTUM BUILD)"
    elif momentum < 0.00:
        return "DISTRIBUTE (TAKING PROFITS)"
    else:
        return "HOLD (RECENTERING)"

clean_df['baxter_action'] = clean_df.apply(baxter_categorize, axis=1)

# 4. Output the Final Actionable Matrix
print("=====================================")
print(" LIVE CATEGORIZATION MATRIX          ")
print("=====================================")
print(clean_df[['ticker', 'harmonized_price', 'momentum_indicator', 'baxter_action']].to_string(index=False))
print("\n[+] AWAITING API TRADE EXECUTION PERMISSION...")
```

```python
import yfinance as yf
import pandas as pd
import time

print("=====================================")
print(" BAXTER BRAIN: LIVE MARKET INGESTION ")
print("=====================================")
time.sleep(1)

# Mapping tickers (Note: Crypto requires the -USD pairing)
tickers = ['SPY', 'BTC-USD', 'NVDA', 'TSLA']

print("[*] Opening outbound tunnel to live market APIs...")
time.sleep(1)
print("[*] Pulling real-time ticker data...")

live_data = []

# Loop through each ticker to pull the live pricing and calculate momentum
for t in tickers:
    stock = yf.Ticker(t)
    hist = stock.history(period="1d")
    
    if not hist.empty:
        current_price = round(hist['Close'].iloc[-1], 2)
        open_price = hist['Open'].iloc[0]
        
        # Calculate live momentum based on intraday percentage change
        momentum = round((current_price - open_price) / open_price * 100, 2) 
        
        # Format the ticker name cleanly for the Grid
        clean_ticker = t.replace('-USD', '')
        
        live_data.append({
            'ticker': clean_ticker, 
            'raw_price': current_price, 
            'momentum_indicator': momentum, 
            'market_noise_ratio': 0.15 # Baseline latency for the Grid to strip
        })

# Build the live dataframe
df_live = pd.DataFrame(live_data)

print("[+] Live market data successfully ingested.\n")
print("=====================================")
print(" RAW LIVE MARKET MATRIX              ")
print("=====================================")
print(df_live.to_string(index=False))
print("\n[+] READY FOR 888 HZ HARMONIZATION...")
```
